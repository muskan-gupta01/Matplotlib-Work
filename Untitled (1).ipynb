```python
# 2. Plot a histogram of Age.
plt.hist(df["Age"],bins=10, color="red")
plt.title("Age_Frequency")
plt.xlabel("Age")
plt.ylabel("Frequency")
plt.show()
```

```python
# 3. Plot histogram of Estimated Salary.
plt.hist(df["EstimatedSalary"],bins=10)
plt.title("Salary Distribution")
plt.xlabel("EstimatedSalary")
plt.ylabel("Frequency")
plt.show()
```

```python
# 4.Plot a bar chart showing numbers of males and females.
gender_count=df["Gender"].value_counts()
plt.bar(gender_count.index, gender_count.values, color=["pink","purple"])
plt.title("Male vs Female Count")
plt.xlabel("Gender")
plt.ylabel("Count")
plt.show()
```

```python
# 5. Create a scatter plot of Age vs Estimated Salary.
plt.scatter(df["Age"],df["EstimatedSalary"])
plt.title("Age VS EstimatedSalary")
plt.xlabel("Age")
plt.ylabel("EstimatedSalary")
plt.show()
```

```python
# 6. Scatter plot with different colors for purchased and not purchased.
colors=["pink" if p==0 else "purple" for p in df["Purchased"]]
plt.scatter(df["Age"],df["EstimatedSalary"],c=colors)
plt.title("Age_EstimatedSalary")
plt.xlabel("Age")
plt.ylabel("EstimatedSalary")
plt.show()
```

```python
# 7. Plot scatter only for users Age>40.
filtered=df[df["Age"]>40]
plt.scatter(filtered["Age"], filtered["EstimatedSalary"])
plt.title("Users Age>40")
plt.show()
```

```python
colors=["red" if age>35 else "blue" for age in df["Age"]]
plt.scatter(df["Age"],df["EstimatedSalary"],c=colors)
plt.title("Age_EstimatedSalary")
plt.xlabel("Age")
plt.ylabel("EstimatedSalary")
plt.show()
```

```python
# 8. Create a boxplot to compare salary for Purchased Vs Not Purchased.
salary_not=df[df["Purchased"]==0]["EstimatedSalary"]
salary_yes=df[df["Purchased"]==1]["EstimatedSalary"]
plt.boxplot([salary_not, salary_yes])
plt.xticks([1,2],["Not Purchased", "Purchased"])
plt.title("Salary Comparsion")
plt.show()
```

```python
# 9. Create two subplots: Age histogram and salary histogram.
plt.figure(figsize=(10,5))
plt.subplot(1,2,1)
plt.hist(df["Age"], color="purple")
plt.title("Age Distribution")
plt.subplot(1,2,2)
plt.hist(df["EstimatedSalary"])
plt.title("Salary Distribution")
plt.tight_layout()
plt.show()
```

```python
# 10. Hoghlight users with Salary >100000 in blue and others in gray.
for i in range(len(df)):
    if df["EstimatedSalary"][i]>100000:
        plt.scatter(df["Age"][i], df["EstimatedSalary"][i], color="blue")
    else:
        plt.scatter(df["Age"][i], df["EstimatedSalary"][i], color="gray")
plt.title("High Salary Highlighted")
plt.xlabel("Age")
plt.ylabel("EstimatedSalary")
plt.show()
```

```python
plt.plot(df["Age"],df["EstimatedSalary"],color="blue",marker="o", markerfacecolor="red")
plt.title("Age vs Salary")
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
gender_count=df["Gender"].value_counts()
plt.pie(gender_count.values,labels=gender_count.index,autopct="%1.1f%%")
plt.title("Gender Distribution")
plt.show()
```

```python
fig,axes=plt.subplots(1,2,figsize=(12,5))
axes[0].hist(df["Age"], bins=20, color="blue", edgecolor="purple")
axes[0].set_title("Age Distribution")
axes[0].set_xlabel("Age")
axes[0].set_ylabel("Frequency")

axes[1].hist(df["EstimatedSalary"], bins=20, color="purple", edgecolor="blue")
axes[1].set_title("EstimatedSalary Distribution")
axes[1].set_xlabel("EstimatedSalary")
axes[1].set_ylabel("Frequency")

plt.tight_layout()
plt.show()
```
